## Formas alternativas de importar

```python
from matematica_exemplo import somar

print(somar(10, 20))
```

```python
from matematica_exemplo import somar as soma

print(soma(1, 1))
```

```python
import matematica_exemplo as mat

print(mat.somar(3, 4))
```

## `if __name__ == "__main__":`

Rode o arquivo diretamente no terminal (fora do notebook) para ver a
diferença: `python3 matematica_exemplo.py`. O bloco dentro do
`if __name__ == "__main__":` só executa nesse caso -- ao importar o
módulo (como fizemos acima), aquele bloco não roda.

```python
import subprocess

resultado = subprocess.run(
    ["python3", "matematica_exemplo.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)
```
