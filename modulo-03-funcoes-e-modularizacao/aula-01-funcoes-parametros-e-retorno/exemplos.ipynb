# Aula 1 — Funções, parâmetros e retorno (exemplos práticos)

```python
def saudacao(nome):
    return f"Olá, {nome}!"

print(saudacao("Ana"))
print(saudacao("Bruno"))
```

## Experimento guiado

Adicione um segundo parâmetro com valor padrão e chame a função de três formas diferentes (posicional, nomeado, usando o padrão).

```python
def saudacao(nome, saudacao_inicial="Olá"):
    return f"{saudacao_inicial}, {nome}!"

print(saudacao("Ana"))
print(saudacao("Ana", "Bom dia"))
print(saudacao(nome="Ana", saudacao_inicial="Oi"))
```

## Retornando múltiplos valores

```python
def calcular_estatisticas(numeros):
    menor = min(numeros)
    maior = max(numeros)
    media = sum(numeros) / len(numeros)
    return menor, maior, media

minimo, maximo, media = calcular_estatisticas([4, 8, 15, 16, 23, 42])
print(minimo, maximo, media)
```

## O erro clássico: confundir print com return

```python
def soma_errada(a, b):
    print(a + b)

resultado = soma_errada(2, 3)
print("resultado capturado:", resultado)  # None -- a função não usou return
```

```python
def soma_certa(a, b):
    return a + b

resultado = soma_certa(2, 3)
print("resultado capturado:", resultado)
```

## Docstrings

```python
def calcular_media(numeros):
    """Retorna a média aritmética de uma lista de números."""
    return sum(numeros) / len(numeros)

print(calcular_media.__doc__)
help(calcular_media)
```

## Mini-desafio resolvido

**Desafio:** função que recebe uma lista de preços e um percentual de desconto, e retorna a lista com os novos preços.

```python
def aplicar_desconto(precos, percentual):
    """Retorna uma nova lista de preços com o percentual de desconto aplicado."""
    return [preco * (1 - percentual / 100) for preco in precos]

precos_originais = [100, 50, 200]
precos_com_desconto = aplicar_desconto(precos_originais, 10)
print(precos_com_desconto)
```
