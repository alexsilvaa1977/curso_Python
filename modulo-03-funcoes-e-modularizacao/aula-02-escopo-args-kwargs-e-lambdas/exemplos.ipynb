# Aula 2 — Escopo, args/kwargs e lambdas (exemplos práticos)

```python
contador_global = 0

def incrementar():
    contador_local = 10
    print("dentro da função, contador_global:", contador_global)
    return contador_local

resultado = incrementar()
print("retorno da função:", resultado)

try:
    print(contador_local)
except NameError as erro:
    print("Erro esperado:", erro)
```

## Evitando `global`: passe e retorne valores

```python
def incrementar(valor_atual):
    return valor_atual + 1

contador = 0
contador = incrementar(contador)
contador = incrementar(contador)
print(contador)
```

## `*args`

```python
def somar_tudo(*numeros):
    print("numeros recebido como tupla:", numeros)
    return sum(numeros)

print(somar_tudo(1, 2))
print(somar_tudo(1, 2, 3, 4, 5))
```

## `**kwargs`

```python
def exibir_dados(**dados):
    for chave, valor in dados.items():
        print(f"{chave}: {valor}")

exibir_dados(nome="Ana", idade=28, cidade="Recife")
```

## Combinando tudo

```python
def registrar_pedido(cliente, *itens, **detalhes):
    print("Cliente:", cliente)
    print("Itens:", itens)
    print("Detalhes:", detalhes)

registrar_pedido("Ana", "camisa", "calça", forma_pagamento="cartão", parcelas=3)
```

## Lambdas

```python
dobro = lambda x: x * 2
print(dobro(5))

pessoas = [{"nome": "Ana", "idade": 28}, {"nome": "Bruno", "idade": 19}]
pessoas_ordenadas = sorted(pessoas, key=lambda p: p["idade"])
print(pessoas_ordenadas)
```

## Mini-desafio resolvido

**Desafio:** função que recebe qualquer quantidade de notas e retorna a média, usando `*args`.

```python
def calcular_media(*notas):
    if not notas:
        return 0
    return sum(notas) / len(notas)

print(calcular_media(8, 9, 7))
print(calcular_media(10))
print(calcular_media())
```
