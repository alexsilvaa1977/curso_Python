## Simulando `pip freeze`

Em um terminal real, `pip freeze > requirements.txt` grava a saída
abaixo em um arquivo. Aqui só exibimos a saída, sem sobrescrever nenhum
arquivo do curso.

````python
resultado = subprocess.run(["pip", "freeze"], capture_output=True, text=True)
print(resultado.stdout[:1000])  # mostrando só os primeiros 1000 caracteres
````

## Mini-desafio resolvido

**Desafio:** ler o `requirements.txt` da raiz do curso e listar, em
Python, quais dependências estão declaradas.

````python
caminho_requirements = "../../requirements.txt"

with open(caminho_requirements, encoding="utf-8") as arquivo:
    linhas = [linha.strip() for linha in arquivo if linha.strip() and not linha.startswith("#")]

print("Dependências declaradas no curso:")
for dependencia in linhas:
    print("-", dependencia)
````
