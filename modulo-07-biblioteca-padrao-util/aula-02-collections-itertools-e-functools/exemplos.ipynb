# Aula 2 — collections, itertools e functools (exemplos práticos)

```python
from collections import Counter

palavras = "python é legal python é útil python vence".split()
contagem = Counter(palavras)
print(contagem)
print(contagem["python"])
print(contagem.most_common(2))
```

## `defaultdict`

```python
from collections import defaultdict

pessoas = [("A", "Ana"), ("B", "Bruno"), ("A", "Carla")]

grupos = defaultdict(list)
for grupo, nome in pessoas:
    grupos[grupo].append(nome)

print(dict(grupos))
```

## `namedtuple`

```python
from collections import namedtuple

Ponto = namedtuple("Ponto", ["x", "y"])

p1 = Ponto(10, 20)
print(p1.x, p1.y)
print(p1)

x, y = p1
print(x, y)
```

## `itertools.chain` e `itertools.product`

```python
from itertools import chain, product

lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
print(list(chain(lista1, lista2)))

cores = ["P", "M"]
tamanhos = ["G", "M"]
print(list(product(cores, tamanhos)))
```

## `itertools.groupby`

```python
from itertools import groupby

numeros = [1, 1, 2, 2, 2, 3, 1, 1]

for chave, grupo in groupby(numeros):
    print(chave, list(grupo))
```

## `functools.reduce`

```python
from functools import reduce

numeros = [1, 2, 3, 4, 5]
produto = reduce(lambda acumulado, atual: acumulado * atual, numeros)
print(produto)
```

## `functools.lru_cache`

```python
from functools import lru_cache
import time

@lru_cache
def calculo_lento(n):
    time.sleep(0.3)
    return n ** 2

inicio = time.perf_counter()
calculo_lento(5)
print(f"primeira chamada: {time.perf_counter() - inicio:.3f}s")

inicio = time.perf_counter()
calculo_lento(5)
print(f"segunda chamada (cache): {time.perf_counter() - inicio:.3f}s")
```

## Mini-desafio resolvido

**Desafio:** usar `@lru_cache` para acelerar o cálculo de Fibonacci.

```python
@lru_cache
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(30))
```
