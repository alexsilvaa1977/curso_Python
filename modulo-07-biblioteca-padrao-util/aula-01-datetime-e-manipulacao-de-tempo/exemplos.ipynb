# Aula 1 — datetime e manipulação de tempo (exemplos práticos)

```python
from datetime import date, datetime, timedelta

hoje = date.today()
print(hoje)
print(hoje.year, hoje.month, hoje.day)

agora = datetime.now()
print(agora)
print(agora.hour, agora.minute, agora.second)
```

## `timedelta`

```python
em_10_dias = hoje + timedelta(days=10)
ha_30_dias = hoje - timedelta(days=30)

print(em_10_dias)
print(ha_30_dias)
```

```python
data_nascimento = date(1995, 3, 15)

diferenca = hoje - data_nascimento
print(diferenca.days, "dias de vida")
print(diferenca.days // 365, "anos aproximadamente")
```

## Comparando datas

```python
data_entrega = date(2026, 12, 25)

if data_entrega > hoje:
    print("Ainda não chegou a data de entrega")
elif data_entrega == hoje:
    print("É hoje!")
else:
    print("A data já passou")
```

## `strftime`: formatando datas

```python
print(agora.strftime("%d/%m/%Y"))
print(agora.strftime("%Y-%m-%d %H:%M"))
print(agora.strftime("%A, %d de %B"))
```

## `strptime`: convertendo texto em data

```python
texto = "15/03/1995"
data_convertida = datetime.strptime(texto, "%d/%m/%Y")
print(data_convertida)
print(data_convertida.date())
```

```python
try:
    datetime.strptime("15-03-1995", "%d/%m/%Y")
except ValueError as erro:
    print("Erro esperado:", erro)
```

## Mini-desafio resolvido

**Desafio:** calcular quantos dias faltam até o fim do ano atual.

```python
hoje = date.today()
fim_do_ano = date(hoje.year, 12, 31)
dias_restantes = (fim_do_ano - hoje).days
print(f"Faltam {dias_restantes} dias para o fim de {hoje.year}.")
```
