# Aula 3 — os, pathlib e sistema de arquivos (exemplos práticos)

Usamos uma pasta temporária como workspace, para não deixar arquivos de
exemplo espalhados pelo repositório do curso.

```python
import tempfile
from pathlib import Path

pasta_temporaria = Path(tempfile.mkdtemp())
print("Workspace desta aula:", pasta_temporaria)

caminho = pasta_temporaria / "dados" / "relatorio.txt"
print(caminho)
```

## Informações sobre um caminho

```python
print(caminho.name)
print(caminho.stem)
print(caminho.suffix)
print(caminho.parent)
print(caminho.parts)
```

## Verificando existência e criando diretórios

```python
print("existe?", caminho.exists())

caminho.parent.mkdir(parents=True, exist_ok=True)
print("pasta criada?", caminho.parent.is_dir())
```

## Lendo e escrevendo com `Path`

```python
caminho.write_text("Reunião às 15h\n", encoding="utf-8")

conteudo = caminho.read_text(encoding="utf-8")
print(conteudo)
print("agora existe?", caminho.exists())
```

## Experimento guiado

Crie mais dois arquivos `.txt` na mesma pasta com `write_text`, e depois liste todos com `.iterdir()` na célula abaixo.

## Listando arquivos com `iterdir` e `glob`

```python
(pasta_temporaria / "dados" / "outro.txt").write_text("outro arquivo", encoding="utf-8")
(pasta_temporaria / "dados" / "notas.csv").write_text("nome,nota\n", encoding="utf-8")

for item in (pasta_temporaria / "dados").iterdir():
    print(item.name, "(pasta)" if item.is_dir() else "(arquivo)")
```

```python
for arquivo_txt in (pasta_temporaria / "dados").glob("*.txt"):
    print(arquivo_txt.name)
```

## Mini-desafio resolvido

**Desafio:** contar quantos arquivos `.txt` existem na pasta.

```python
quantidade = len(list((pasta_temporaria / "dados").glob("*.txt")))
print(f"Existem {quantidade} arquivos .txt na pasta.")
```

## Limpando o workspace temporário

```python
import shutil

shutil.rmtree(pasta_temporaria)
print("Workspace temporário removido.")
```
