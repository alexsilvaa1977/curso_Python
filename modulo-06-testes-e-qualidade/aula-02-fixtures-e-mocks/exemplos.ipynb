## Entendendo o mock por dentro

```python
from unittest.mock import Mock

resposta_falsa = Mock()
resposta_falsa.json.return_value = {"temperatura": 25}

print(resposta_falsa.json())          # {'temperatura': 25} -- comportamento configurado
print(resposta_falsa.json.call_count)  # 1 -- o Mock também registra quantas vezes foi chamado
```

## Mini-desafio resolvido

**Desafio:** usar `monkeypatch` (via um teste real) para simular uma função `obter_data_atual()` e testar um comportamento que depende da data.

```python
codigo_teste = '''
import datetime

def obter_data_atual():
    return datetime.date.today()

def eh_fim_de_semana():
    return obter_data_atual().weekday() >= 5


def test_eh_fim_de_semana_no_sabado(monkeypatch):
    import datetime as datetime_module

    class DataFalsaSabado(datetime_module.date):
        @classmethod
        def today(cls):
            return cls(2024, 1, 6)  # um sábado

    monkeypatch.setattr(datetime_module, "date", DataFalsaSabado)
    assert eh_fim_de_semana() is True
'''

with open("test_temporario_fim_de_semana.py", "w", encoding="utf-8") as arquivo:
    arquivo.write(codigo_teste)

resultado = subprocess.run(
    ["python3", "-m", "pytest", "-v", "test_temporario_fim_de_semana.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)

import os
os.remove("test_temporario_fim_de_semana.py")
```
