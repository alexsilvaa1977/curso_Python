## Um teste com cobertura, mas sem verificação real

```python
codigo_teste_ruim = '''
from calculadora import somar

def test_ruim_mas_com_cobertura():
    somar(2, 3)  # executa a linha, mas nao verifica nada
'''

with open("test_temporario_ruim.py", "w", encoding="utf-8") as arquivo:
    arquivo.write(codigo_teste_ruim)

resultado = subprocess.run(
    ["python3", "-m", "pytest", "--cov=calculadora", "-v", "test_temporario_ruim.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)

import os
os.remove("test_temporario_ruim.py")
```

Note que esse teste passa e até conta como "cobertura" da linha de
`somar`, mas não verificaria nada se `somar` estivesse quebrada (ex.: se
alguém trocasse `return a + b` por `return a - b`, esse teste continuaria
passando). Cobertura mede execução, não qualidade da verificação.

## Mini-desafio resolvido

**Desafio:** completar os testes até atingir 100% de cobertura em `calculadora.py`.

```python
codigo_teste_completo = '''
import pytest
from calculadora import somar, dividir, eh_par

def test_somar():
    assert somar(2, 3) == 5

def test_dividir():
    assert dividir(10, 2) == 5

def test_eh_par_com_numero_par():
    assert eh_par(4) is True

def test_eh_par_com_numero_impar():
    assert eh_par(3) is False
'''

with open("test_temporario_completo.py", "w", encoding="utf-8") as arquivo:
    arquivo.write(codigo_teste_completo)

resultado = subprocess.run(
    ["python3", "-m", "pytest", "--cov=calculadora", "--cov-report=term-missing",
     "-v", "test_temporario_completo.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)

import os
os.remove("test_temporario_completo.py")
```
