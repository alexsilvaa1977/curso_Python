## `Optional` e tipos genéricos modernos

```python
from typing import Optional

def buscar_usuario(id_usuario: int) -> Optional[str]:
    usuarios = {1: "Ana", 2: "Bruno"}
    return usuarios.get(id_usuario)

print(buscar_usuario(1))
print(buscar_usuario(999))
```

```python
def calcular_media(notas: list[float]) -> float:
    return sum(notas) / len(notas)

def contar_palavras(texto: str) -> dict[str, int]:
    contagem: dict[str, int] = {}
    for palavra in texto.split():
        contagem[palavra] = contagem.get(palavra, 0) + 1
    return contagem

print(calcular_media([8.0, 9.0, 7.0]))
print(contar_palavras("python é legal python é útil"))
```

## Rodando `mypy` de verdade sobre um arquivo com erro de tipo

```python
import subprocess

codigo_com_erro = '''def somar(a: int, b: int) -> int:
    return a + b

resultado = somar(2, "3")
'''

with open("_exemplo_mypy_erro.py", "w", encoding="utf-8") as arquivo:
    arquivo.write(codigo_com_erro)

resultado = subprocess.run(
    ["python3", "-m", "mypy", "_exemplo_mypy_erro.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)

import os
os.remove("_exemplo_mypy_erro.py")
```

## Mini-desafio resolvido

**Desafio:** tipar uma função que recebe uma lista de pessoas (dicionários) e retorna apenas os nomes.

```python
def extrair_nomes(pessoas: list[dict[str, object]]) -> list[str]:
    return [str(pessoa["nome"]) for pessoa in pessoas]

pessoas = [{"nome": "Ana", "idade": 28}, {"nome": "Bruno", "idade": 34}]
print(extrair_nomes(pessoas))
```
