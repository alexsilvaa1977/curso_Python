# Aula 1 — Introdução a testes com pytest (exemplos práticos)

Esta pasta já contém `calculadora.py` (o código sendo testado) e
`test_calculadora.py` (os testes) ao lado deste notebook. Vamos rodar o
`pytest` de verdade sobre eles, via `subprocess`.

```python
from calculadora import somar, dividir

assert somar(2, 3) == 5
print("assert simples passou sem erro")
```

```python
try:
    assert somar(2, 3) == 10
except AssertionError as erro:
    print("AssertionError capturado (esperado):", erro)
```

## Rodando o `pytest` real sobre `test_calculadora.py`

```python
import subprocess

resultado = subprocess.run(
    ["python3", "-m", "pytest", "-v", "test_calculadora.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)
print(resultado.stderr)
```

## `pytest.raises`

```python
import pytest

with pytest.raises(ZeroDivisionError):
    dividir(10, 0)

print("A exceção esperada realmente aconteceu -- nenhum erro chegou até aqui.")
```

## Mini-desafio resolvido

**Desafio:** escrever e rodar um teste para uma função `eh_par(numero)` ainda não testada.

```python
def eh_par(numero):
    return numero % 2 == 0

# Testando diretamente aqui no notebook com assert (fora do arquivo test_*.py,
# só para fins de demonstração rápida)
assert eh_par(4) is True
assert eh_par(7) is False
print("Todos os asserts passaram.")
```
