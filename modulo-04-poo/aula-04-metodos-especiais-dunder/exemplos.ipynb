# Aula 4 — Métodos especiais (dunder methods) (exemplos práticos)

```python
class PessoaSemStr:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

ana_sem_str = PessoaSemStr("Ana", 28)
print(ana_sem_str)  # representação padrão, pouco útil
```

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def __str__(self):
        return f"{self.nome} ({self.idade} anos)"

    def __repr__(self):
        return f"Pessoa(nome={self.nome!r}, idade={self.idade})"

ana = Pessoa("Ana", 28)
print(ana)
print(str(ana))
print(repr(ana))
print([ana, ana])
```

## `__eq__`

```python
class PontoSemEq:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = PontoSemEq(1, 2)
p2 = PontoSemEq(1, 2)
print("sem __eq__:", p1 == p2)
```

```python
class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, outro):
        return self.x == outro.x and self.y == outro.y

    def __repr__(self):
        return f"Ponto({self.x}, {self.y})"

p1 = Ponto(1, 2)
p2 = Ponto(1, 2)
print("com __eq__:", p1 == p2)
```

## `__len__`

```python
class Carrinho:
    def __init__(self):
        self.itens = []

    def adicionar(self, item):
        self.itens.append(item)

    def __len__(self):
        return len(self.itens)

carrinho = Carrinho()
carrinho.adicionar("Teclado")
carrinho.adicionar("Mouse")
print(len(carrinho))
```

## Mini-desafio resolvido

**Desafio:** classe `Dinheiro` com `__eq__` e `__repr__`, comparando valores em centavos para evitar problemas de arredondamento com float.

```python
class Dinheiro:
    def __init__(self, reais, centavos=0):
        self.total_centavos = reais * 100 + centavos

    def __eq__(self, outro):
        return self.total_centavos == outro.total_centavos

    def __repr__(self):
        reais, centavos = divmod(self.total_centavos, 100)
        return f"Dinheiro(R$ {reais},{centavos:02d})"

preco_a = Dinheiro(19, 90)
preco_b = Dinheiro(19, 90)
preco_c = Dinheiro(20, 0)

print(preco_a)
print(preco_a == preco_b)
print(preco_a == preco_c)
```
