# Aula 2 — Herança e polimorfismo (exemplos práticos)

```python
class Animal:
    def __init__(self, nome):
        self.nome = nome

    def emitir_som(self):
        return "..."

class Cachorro(Animal):
    def emitir_som(self):
        return "Woof!"

class Gato(Animal):
    def emitir_som(self):
        return "Miau!"

rex = Cachorro("Rex")
print(rex.nome)
print(rex.emitir_som())
```

## `super()`

```python
class Cachorro(Animal):
    def __init__(self, nome, raca):
        super().__init__(nome)
        self.raca = raca

rex = Cachorro("Rex", "Labrador")
print(rex.nome, rex.raca)
```

## Polimorfismo

```python
class Gato(Animal):
    def __init__(self, nome):
        super().__init__(nome)

    def emitir_som(self):
        return "Miau!"

animais = [Cachorro("Rex", "Labrador"), Gato("Mimi")]

for animal in animais:
    print(f"{animal.nome}: {animal.emitir_som()}")
```

## Verificando tipo e hierarquia

```python
print(isinstance(rex, Cachorro))
print(isinstance(rex, Animal))
print(isinstance(rex, Gato))
print(type(rex))
print(Cachorro.__bases__)
```

## Herança múltipla

```python
class Nadador:
    def nadar(self):
        return "nadando..."

class Voador:
    def voar(self):
        return "voando..."

class Pato(Nadador, Voador):
    pass

pato = Pato()
print(pato.nadar(), pato.voar())
```

## Mini-desafio resolvido

**Desafio:** hierarquia `Funcionario` -> `Gerente`, onde `Gerente` tem um bônus adicional no salário.

```python
class Funcionario:
    def __init__(self, nome, salario_base):
        self.nome = nome
        self.salario_base = salario_base

    def calcular_salario(self):
        return self.salario_base

class Gerente(Funcionario):
    def __init__(self, nome, salario_base, bonus):
        super().__init__(nome, salario_base)
        self.bonus = bonus

    def calcular_salario(self):
        return super().calcular_salario() + self.bonus

funcionarios = [
    Funcionario("Ana", 3000),
    Gerente("Bruno", 4000, 1500),
]

for funcionario in funcionarios:
    print(funcionario.nome, funcionario.calcular_salario())
```
