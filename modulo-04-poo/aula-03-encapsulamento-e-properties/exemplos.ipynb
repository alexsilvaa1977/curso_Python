# Aula 3 — Encapsulamento e properties (exemplos práticos)

```python
class ContaBancaria:
    def __init__(self, saldo_inicial):
        self.__saldo = saldo_inicial

conta = ContaBancaria(100)

try:
    print(conta.__saldo)
except AttributeError as erro:
    print("Erro esperado:", erro)

print(conta._ContaBancaria__saldo)  # funciona, mas ninguém deveria escrever isso
```

## `@property` (somente leitura)

```python
class ContaBancaria:
    def __init__(self, titular, saldo_inicial):
        self.titular = titular
        self._saldo = saldo_inicial

    @property
    def saldo(self):
        return self._saldo

conta = ContaBancaria("Ana", 100)
print(conta.saldo)
```

## Adicionando um `@saldo.setter` com validação

```python
class ContaBancaria:
    def __init__(self, titular, saldo_inicial):
        self.titular = titular
        self._saldo = saldo_inicial

    @property
    def saldo(self):
        return self._saldo

    @saldo.setter
    def saldo(self, novo_valor):
        if novo_valor < 0:
            raise ValueError("Saldo não pode ser negativo")
        self._saldo = novo_valor

conta = ContaBancaria("Ana", 100)
conta.saldo = 200
print(conta.saldo)

try:
    conta.saldo = -50
except ValueError as erro:
    print("Erro esperado:", erro)
```

## Property calculada (sem setter)

```python
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura

    @property
    def area(self):
        return self.largura * self.altura

r = Retangulo(4, 5)
print(r.area)

try:
    r.area = 100
except AttributeError as erro:
    print("Erro esperado:", erro)
```

## Mini-desafio resolvido

**Desafio:** classe `Produto` com `preco` validado (não pode ser negativo) via property.

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco   # já passa pelo setter abaixo

    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, valor):
        if valor < 0:
            raise ValueError("Preço não pode ser negativo")
        self._preco = valor

produto = Produto("Teclado", 150)
print(produto.preco)

try:
    produto.preco = -10
except ValueError as erro:
    print("Erro esperado:", erro)
```
