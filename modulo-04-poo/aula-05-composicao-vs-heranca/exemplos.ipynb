# Aula 5 — Composição vs. herança (exemplos práticos)

```python
class Motor:
    def __init__(self, potencia_cv):
        self.potencia_cv = potencia_cv

    def ligar(self):
        return "Motor ligado"

class Carro:
    def __init__(self, modelo, motor):
        self.modelo = modelo
        self.motor = motor

    def ligar(self):
        return f"{self.modelo}: {self.motor.ligar()}"

motor_v6 = Motor(300)
carro = Carro("Sedan X", motor_v6)
print(carro.ligar())
```

## Herança mal aplicada: o sinal de alerta

```python
class Voador:
    def voar(self):
        return "voando..."

class Passaro(Voador):
    pass

class Pinguim(Passaro):
    def voar(self):
        raise NotImplementedError("Pinguins não voam!")

aguia = Passaro()
pinguim = Pinguim()

print(aguia.voar())

try:
    pinguim.voar()
except NotImplementedError as erro:
    print("Erro esperado (sinal de design problemático):", erro)
```

## Reescrevendo com composição

```python
class CapacidadeVoo:
    def voar(self):
        return "voando..."

class Passaro:
    def __init__(self, nome, capacidade_voo=None):
        self.nome = nome
        self.capacidade_voo = capacidade_voo

    def tentar_voar(self):
        if self.capacidade_voo is None:
            return f"{self.nome} não sabe voar."
        return f"{self.nome}: {self.capacidade_voo.voar()}"

aguia = Passaro("Águia", CapacidadeVoo())
pinguim = Passaro("Pinguim")

print(aguia.tentar_voar())
print(pinguim.tentar_voar())
```

## Mini-desafio resolvido

**Desafio:** um `Pedido` que TEM uma lista de `ItemPedido` (composição), calculando o total.

```python
class ItemPedido:
    def __init__(self, nome, preco, quantidade):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade

    def subtotal(self):
        return self.preco * self.quantidade

class Pedido:
    def __init__(self, cliente):
        self.cliente = cliente
        self.itens = []

    def adicionar_item(self, item):
        self.itens.append(item)

    def total(self):
        return sum(item.subtotal() for item in self.itens)

pedido = Pedido("Ana")
pedido.adicionar_item(ItemPedido("Teclado", 150, 1))
pedido.adicionar_item(ItemPedido("Mouse", 45, 2))

print(pedido.cliente, "- total:", pedido.total())
```
