# Aula 2 — Introdução ao Flask (exemplos práticos)

Esta pasta contém `app.py` (a API) e `test_app.py` (os testes). Em vez
de rodar `app.run()` (que bloquearia o notebook esperando requisições
para sempre), vamos usar o `test_client()` do próprio Flask -- a forma
recomendada de testar uma aplicação Flask.

```python
from app import app

cliente = app.test_client()

resposta = cliente.get("/tarefas")
print(resposta.status_code)
print(resposta.get_json())
```

## Buscando uma tarefa específica

```python
resposta = cliente.get("/tarefas/1")
print(resposta.status_code)
print(resposta.get_json())

resposta = cliente.get("/tarefas/999")
print(resposta.status_code)
print(resposta.get_json())
```

## Criando uma tarefa com `POST`

```python
resposta = cliente.post("/tarefas", json={"titulo": "Praticar exercícios do módulo 8"})
print(resposta.status_code)
print(resposta.get_json())
```

## Experimento guiado

Crie mais duas tarefas com `POST` e depois rode `cliente.get("/tarefas")` de novo para ver todas elas na lista.

## Atualizando e removendo

```python
resposta = cliente.put("/tarefas/1", json={"concluida": True})
print(resposta.status_code, resposta.get_json())

resposta = cliente.delete("/tarefas/1")
print(resposta.status_code)

resposta = cliente.get("/tarefas/1")
print(resposta.status_code)  # 404 -- a tarefa 1 foi removida
```

## Rodando a suíte de testes real com `pytest`

```python
import subprocess

resultado = subprocess.run(
    ["python3", "-m", "pytest", "-v", "test_app.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)
```

## Mini-desafio resolvido

**Desafio:** contar quantas tarefas estão concluídas, usando a própria API.

```python
resposta = cliente.get("/tarefas")
tarefas = resposta.get_json()
concluidas = sum(1 for tarefa in tarefas if tarefa["concluida"])
print(f"{concluidas} de {len(tarefas)} tarefas concluídas.")
```
