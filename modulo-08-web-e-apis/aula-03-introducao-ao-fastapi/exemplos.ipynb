# Aula 3 — Introdução ao FastAPI (exemplos práticos)

Esta pasta contém `main.py` (a API) e `test_main.py` (os testes). Assim
como fizemos com o Flask, vamos usar o `TestClient` do FastAPI em vez de
rodar `uvicorn` de verdade -- ideal para explorar e testar sem bloquear
o notebook.

```python
from fastapi.testclient import TestClient
from main import app

cliente = TestClient(app)

resposta = cliente.get("/tarefas")
print(resposta.status_code)
print(resposta.json())
```

## Validação automática com Pydantic

```python
resposta = cliente.get("/tarefas/abc")
print(resposta.status_code)
print(resposta.json())
```

```python
resposta = cliente.post("/tarefas", json={})
print(resposta.status_code)
print(resposta.json())
```

## Experimento guiado

Envie `cliente.post("/tarefas", json={"titulo": 123})` (um número em
vez de texto) e observe a mensagem de erro do Pydantic explicando qual
campo e por que a validação falhou.

## Criando e removendo tarefas

```python
resposta = cliente.post("/tarefas", json={"titulo": "Praticar FastAPI"})
print(resposta.status_code)
tarefa_criada = resposta.json()
print(tarefa_criada)
```

```python
resposta = cliente.delete(f"/tarefas/{tarefa_criada['id']}")
print(resposta.status_code)

resposta = cliente.get(f"/tarefas/{tarefa_criada['id']}")
print(resposta.status_code)
```

## Rodando a suíte de testes real com `pytest`

```python
import subprocess

resultado = subprocess.run(
    ["python3", "-m", "pytest", "-v", "test_main.py"],
    capture_output=True,
    text=True,
)
print(resultado.stdout)
```

## Mini-desafio resolvido

**Desafio:** usar o schema OpenAPI gerado automaticamente pelo FastAPI para listar todas as rotas da API, sem ler o código-fonte.

```python
schema = app.openapi()
for caminho, operacoes in schema["paths"].items():
    for metodo in operacoes:
        print(metodo.upper(), caminho)
```
