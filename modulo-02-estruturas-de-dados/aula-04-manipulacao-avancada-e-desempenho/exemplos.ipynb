## `zip()`

```python
nomes = ["Ana", "Bruno", "Carla"]
idades = [28, 34, 25]

for nome, idade in zip(nomes, idades):
    print(nome, idade)

pessoas_dict = dict(zip(nomes, idades))
print(pessoas_dict)
```

## `map()` e `filter()` vs. comprehension

```python
numeros = [1, 2, 3, 4, 5]

dobrados_map = list(map(lambda n: n * 2, numeros))
dobrados_comp = [n * 2 for n in numeros]
print(dobrados_map == dobrados_comp)

pares_filter = list(filter(lambda n: n % 2 == 0, numeros))
pares_comp = [n for n in numeros if n % 2 == 0]
print(pares_filter == pares_comp)
```

## Lista vs. set: comparando velocidade de busca
