# Aula 1 — Listas e tuplas (exemplos práticos)

```python
frutas = ["maçã", "banana", "uva"]
print(frutas[0])
print(frutas[-1])
print(frutas[1:])
```

## Experimento guiado

Adicione mais frutas com `.append()` e depois `.insert()` em outra posição, imprimindo a lista depois de cada operação.

```python
frutas.append("laranja")
print(frutas)

frutas.insert(1, "pera")
print(frutas)

frutas.remove("banana")
print(frutas)

ultimo = frutas.pop()
print("removido:", ultimo, "| lista agora:", frutas)
```

```python
numeros = [5, 3, 1, 4, 2]

# sorted() -- não altera a lista original
numeros_ordenados = sorted(numeros)
print("original:", numeros)
print("nova, ordenada:", numeros_ordenados)

# .sort() -- altera a lista original, no lugar
numeros.sort()
print("original após .sort():", numeros)
```

## `enumerate` para percorrer com índice

```python
frutas = ["maçã", "banana", "uva"]
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

## Tuplas e desempacotamento

## Mini-desafio resolvido

**Desafio:** encontrar o maior e o menor valor de uma lista sem usar `max()`/`min()`.

```python
numeros = [7, 2, 9, 4, 1, 8]

maior = numeros[0]
menor = numeros[0]

for numero in numeros:
    if numero > maior:
        maior = numero
    if numero < menor:
        menor = numero

print("maior:", maior, "| menor:", menor)
```
