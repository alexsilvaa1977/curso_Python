# Aula 2 — Dicionários e sets (exemplos práticos)

```python
pessoa = {
    "nome": "Ana",
    "idade": 28,
    "cidade": "Recife",
}
print(pessoa["nome"])
print(pessoa["idade"])
```

## Experimento guiado

Adicione uma chave nova (`profissao`) e remova `cidade`. Depois tente acessar `pessoa["telefone"]` e veja o erro.

```python
pessoa["idade"] = 29
pessoa["profissao"] = "dev"
del pessoa["cidade"]
print(pessoa)

print(pessoa.get("telefone"))
print(pessoa.get("telefone", "não informado"))
```

```python
try:
    print(pessoa["telefone"])
except KeyError as erro:
    print("Erro esperado:", erro)
```

## Percorrendo dicionários

```python
for chave, valor in pessoa.items():
    print(chave, "->", valor)
```

## Dicionários aninhados

```python
alunos = {
    "aluno1": {"nome": "Ana", "notas": [8, 9, 7]},
    "aluno2": {"nome": "Bruno", "notas": [6, 7, 5]},
}

print(alunos["aluno1"]["nome"])
print(alunos["aluno1"]["notas"][0])

for chave_aluno, dados in alunos.items():
    media = sum(dados["notas"]) / len(dados["notas"])
    print(chave_aluno, dados["nome"], "média:", media)
```

## Sets e operações de conjunto

```python
frutas = {"maçã", "banana", "maçã", "uva"}
print(frutas)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("união:", a | b)
print("interseção:", a & b)
print("diferença a-b:", a - b)
print("diferença simétrica:", a ^ b)
```
