# Aula 3 — Comprehensions (exemplos práticos)

```python
quadrados_for = []
for numero in range(10):
    quadrados_for.append(numero ** 2)

quadrados_comp = [numero ** 2 for numero in range(10)]

print(quadrados_for)
print(quadrados_comp)
print(quadrados_for == quadrados_comp)
```

## Experimento guiado

Mude a condição `% 2 == 0` para `% 3 == 0` e veja a lista mudar.

```python
pares = [numero for numero in range(20) if numero % 2 == 0]
print(pares)
```

```python
resultado = ["par" if n % 2 == 0 else "ímpar" for n in range(5)]
print(resultado)
```

## Dict e set comprehension

```python
nomes = ["ana", "bruno", "carla"]
tamanhos = {nome: len(nome) for nome in nomes}
print(tamanhos)

precos = {"maçã": 2.5, "banana": 1.8, "uva": 6.0}
precos_com_desconto = {produto: round(preco * 0.9, 2) for produto, preco in precos.items()}
print(precos_com_desconto)

palavras = ["python", "java", "python", "go", "java"]
linguagens_unicas = {p.upper() for p in palavras}
print(linguagens_unicas)
```

## Achatando uma lista de listas

```python
matriz = [[1, 2, 3], [4, 5, 6]]
achatada = [numero for linha in matriz for numero in linha]
print(achatada)
```

## Mini-desafio resolvido

**Desafio:** a partir de uma lista de nomes, criar um dicionário `{nome: "curto" ou "longo"}` (curto = até 4 letras).

```python
nomes = ["Ana", "Bruno", "Ig", "Fernanda", "Zeca"]
classificacao = {nome: ("curto" if len(nome) <= 4 else "longo") for nome in nomes}
print(classificacao)
```
