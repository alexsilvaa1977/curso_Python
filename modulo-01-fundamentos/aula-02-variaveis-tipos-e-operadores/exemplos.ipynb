## Experimento guiado

Mude os valores acima e rode de novo. Depois, tente atribuir um texto a `idade` e veja o `type()` mudar.

```python
idade_texto = input("Sua idade: ")
idade = int(idade_texto)
print(f"No ano que vem você terá {idade + 1} anos.")
```

```python
# Cuidado com esta conta -- ela dá erro de propósito, para você ver o traceback
try:
    resultado = "idade: " + 25
except TypeError as erro:
    print("Erro esperado:", erro)

# Forma correta:
resultado = "idade: " + str(25)
print(resultado)
```

## Operadores aritméticos na prática

```python
a = 7
b = 3
print("soma:", a + b)
print("subtração:", a - b)
print("multiplicação:", a * b)
print("divisão:", a / b)
print("divisão inteira:", a // b)
print("resto:", a % b)
print("potência:", a ** 2)
```

## Par ou ímpar usando `%`

```python
numero = 17
if numero % 2 == 0:
    print("par")
else:
    print("ímpar")
```

## Mini-desafio resolvido

**Desafio:** calcular se uma pessoa pode dirigir (maior de 18 e com carteira).

```python
idade = 20
tem_carteira = True

pode_dirigir = idade >= 18 and tem_carteira
print("Pode dirigir?", pode_dirigir)

# Testando o caso contrário
tem_carteira = False
pode_dirigir = idade >= 18 and tem_carteira
print("Pode dirigir?", pode_dirigir)
```
