# Aula 4 — Strings e formatação (exemplos práticos)

```python
palavra = "Python"
print(palavra[0])
print(palavra[-1])
print(len(palavra))
```

## Experimento guiado

Mude os índices abaixo (`0:3`, `::-1`, `::2`) e observe o resultado.

```python
print(palavra[0:3])
print(palavra[:3])
print(palavra[3:])
print(palavra[::-1])
print(palavra[::2])
```

```python
# Strings são imutáveis -- isto dá erro de propósito
try:
    palavra[0] = "J"
except TypeError as erro:
    print("Erro esperado:", erro)

nova_palavra = "J" + palavra[1:]
print(nova_palavra)
```

## Métodos de string

```python
texto = "  Curso de Python  "

print(repr(texto.strip()))
print(texto.lower())
print(texto.upper())
print(texto.replace("Python", "Django"))
print(texto.split())
print("-".join(["a", "b", "c"]))
print(texto.strip().startswith("Curso"))
print("python" in texto.lower())
```

## f-strings

```python
nome = "Ana"
idade = 28

print(f"{nome} tem {idade} anos.")
print(f"No ano que vem, {nome} terá {idade + 1} anos.")

preco = 19.9
print(f"Preço: R$ {preco:.2f}")

percentual = 0.256
print(f"{percentual:.1%}")
```
