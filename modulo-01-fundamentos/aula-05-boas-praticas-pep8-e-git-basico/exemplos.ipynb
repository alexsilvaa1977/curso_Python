## Comparando estilos

As duas células abaixo fazem exatamente a mesma coisa. Note a diferença de legibilidade.

````python
# Estilo ruim (funciona, mas viola PEP 8)
def calc(a,b,c):
    x=a+b*c
    return x
print(calc(1,2,3))
````

````python
# Estilo recomendado (PEP 8): espaços, nomes descritivos
def calcular_total(preco_base, quantidade, taxa_extra):
    total = preco_base + quantidade * taxa_extra
    return total

print(calcular_total(1, 2, 3))
````

## Convenção de nomes

````python
MAX_TENTATIVAS = 3

class ContaBancaria:
    pass

def sacar_dinheiro(valor):
    return valor

print(MAX_TENTATIVAS, ContaBancaria, sacar_dinheiro(10))
````

## Mini-desafio resolvido

**Desafio:** dado o código malformatado abaixo, reescreva seguindo PEP 8.

````python
# Antes
def soma_lista(lista):
    total=0
    for x in lista:
        total=total+x
    return total
````

````python
# Depois (PEP 8: nomes claros, espaços em operadores, uso de += )
def somar_lista(numeros):
    total = 0
    for numero in numeros:
        total += numero
    return total

print(somar_lista([1, 2, 3, 4]))
````
