## Descobrindo o tipo de um valor

A função `type()` mostra o tipo de qualquer valor — vamos usar bastante
isso na próxima aula, quando falarmos de tipos de dados.

```python
print(type("texto"))
print(type(10))
print(type(3.14))
```

## Mini-desafio resolvido

**Desafio:** peça o nome e a idade da pessoa, e exiba uma frase completa
com os dois dados.

Resolução passo a passo:

```python
nome = input("Nome: ")
idade = input("Idade: ")

mensagem = f"{nome} tem {idade} anos."
print(mensagem)
```

Note que `idade` chega como **texto** (string), mesmo que a pessoa digite
só números — `input()` sempre devolve string. Na próxima aula vamos ver
como converter isso para número quando precisarmos calcular com ele.
