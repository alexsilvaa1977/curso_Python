# Aula 3 — Manipulação de arquivos texto e CSV (exemplos práticos)

Para não deixar arquivos de exemplo espalhados pelo repositório do
curso, usamos uma pasta temporária (`tempfile`) como "workspace" desta
aula -- na prática, você usaria caminhos normais do seu projeto.

```python
import tempfile
import os

pasta_temporaria = tempfile.mkdtemp()
print("Workspace desta aula:", pasta_temporaria)

caminho_notas = os.path.join(pasta_temporaria, "notas.txt")
```

## Escrevendo e lendo um arquivo de texto

```python
with open(caminho_notas, "w", encoding="utf-8") as arquivo:
    arquivo.write("Ana: 8.5\n")
    arquivo.write("Bruno: 7.0\n")
    arquivo.write("Carla: 9.2\n")

with open(caminho_notas, "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
```

## Lendo linha por linha

```python
with open(caminho_notas, "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        print(linha.strip())
```

## Modo `"a"` (append) -- adicionando sem apagar

```python
with open(caminho_notas, "a", encoding="utf-8") as arquivo:
    arquivo.write("Diego: 6.5\n")

with open(caminho_notas, "r", encoding="utf-8") as arquivo:
    print(arquivo.read())
```

## Tratando arquivo inexistente

```python
caminho_inexistente = os.path.join(pasta_temporaria, "nao_existe.txt")

try:
    with open(caminho_inexistente, "r") as arquivo:
        conteudo = arquivo.read()
except FileNotFoundError:
    print("Arquivo não encontrado.")
```

## Trabalhando com CSV

```python
import csv

caminho_csv = os.path.join(pasta_temporaria, "produtos.csv")

with open(caminho_csv, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["nome", "preco", "estoque"])
    escritor.writerow(["Teclado", 150.0, 20])
    escritor.writerow(["Mouse", 45.0, 50])

with open(caminho_csv, "r", encoding="utf-8") as arquivo:
    leitor = csv.reader(arquivo)
    for linha in leitor:
        print(linha)
```

## `csv.DictReader` e `csv.DictWriter`

```python
with open(caminho_csv, "r", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    for linha in leitor:
        print(linha["nome"], "->", linha["preco"])
```

## Mini-desafio resolvido

**Desafio:** ler o CSV e calcular o valor total em estoque (`preco * estoque` de cada produto).

```python
with open(caminho_csv, "r", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    total = sum(float(linha["preco"]) * int(linha["estoque"]) for linha in leitor)

print(f"Valor total em estoque: R$ {total:.2f}")
```

## Limpando o workspace temporário

```python
import shutil

shutil.rmtree(pasta_temporaria)
print("Workspace temporário removido.")
```
