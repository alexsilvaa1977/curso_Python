# Aula 1 — Exceções: try/except/finally (exemplos práticos)

```python
try:
    idade = int(input("Digite sua idade: "))
    print(f"Você tem {idade} anos.")
except ValueError:
    print("Isso não é um número válido.")
```

## Múltiplos `except`

```python
try:
    numero = int(input("Digite um número: "))
    resultado = 100 / numero
    print(resultado)
except ValueError:
    print("Isso não é um número.")
except ZeroDivisionError:
    print("Não é possível dividir por zero.")
```

## Por que evitar `except:` genérico

```python
def dividir(a, b):
    return a / b

# Ruim: captura literalmente tudo, esconde o bug de digitação abaixo
try:
    resultado = dividr(10, 2)  # nome errado de propósito -- NameError
except:
    print("Deu erro (mas não sabemos qual!)")
```

```python
# Melhor: específico o suficiente para não esconder o NameError
try:
    resultado = dividr(10, 2)  # mesmo erro de digitação
except ZeroDivisionError:
    print("Divisão por zero")
except NameError as erro:
    print("Erro real, capturado corretamente:", erro)
```

## `else` e `finally`

```python
try:
    numero = int("10")
except ValueError:
    print("Entrada inválida.")
else:
    print(f"Você digitou {numero}, que ao quadrado é {numero ** 2}.")
finally:
    print("Bloco finally sempre executa.")
```

## `raise`: levantando exceções manualmente

```python
def sacar(saldo, valor):
    if valor > saldo:
        raise ValueError("Saldo insuficiente")
    return saldo - valor

try:
    sacar(100, 500)
except ValueError as erro:
    print("Operação negada:", erro)

print("Saldo após saque válido:", sacar(100, 30))
```

## Mini-desafio resolvido

**Desafio:** função que lê uma lista de textos e converte para números, contando quantos falharam.

```python
def converter_para_numeros(textos):
    numeros = []
    falhas = 0
    for texto in textos:
        try:
            numeros.append(int(texto))
        except ValueError:
            falhas += 1
    return numeros, falhas

entradas = ["10", "20", "abc", "30", "xyz"]
numeros, falhas = converter_para_numeros(entradas)
print("Números válidos:", numeros)
print("Falhas:", falhas)
```
