# Aula 4 — Serialização: JSON e pickle (exemplos práticos)

```python
import json

dados = {
    "nome": "Ana",
    "idade": 28,
    "linguagens": ["Python", "SQL"],
    "ativo": True,
}

texto_json = json.dumps(dados)
print(texto_json)
```

```python
texto_json = '{"nome": "Bruno", "idade": 34}'
dados_carregados = json.loads(texto_json)
print(dados_carregados["nome"], dados_carregados["idade"])
```

## Formatação legível

```python
texto_formatado = json.dumps(dados, indent=2, ensure_ascii=False)
print(texto_formatado)
```

## Salvando e carregando JSON em arquivo

```python
import tempfile
import os

pasta_temporaria = tempfile.mkdtemp()
caminho_json = os.path.join(pasta_temporaria, "pessoa.json")

with open(caminho_json, "w", encoding="utf-8") as arquivo:
    json.dump(dados, arquivo, indent=2, ensure_ascii=False)

with open(caminho_json, "r", encoding="utf-8") as arquivo:
    dados_lidos = json.load(arquivo)

print(dados_lidos)
```

## Limitação do JSON com objetos customizados

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

ana = Pessoa("Ana", 28)

try:
    json.dumps(ana)
except TypeError as erro:
    print("Erro esperado:", erro)

dados_pessoa = {"nome": ana.nome, "idade": ana.idade}
print(json.dumps(dados_pessoa))
```

## `pickle`: serializando o objeto Python diretamente

```python
import pickle

caminho_pickle = os.path.join(pasta_temporaria, "pessoa.pkl")

with open(caminho_pickle, "wb") as arquivo:
    pickle.dump(ana, arquivo)

with open(caminho_pickle, "rb") as arquivo:
    ana_recuperada = pickle.load(arquivo)

print(ana_recuperada.nome, ana_recuperada.idade)
```

## Experimento guiado

Tente abrir o arquivo `.pkl` com `open(caminho_pickle, "r")` (modo texto,
sem o `b`) e leia o conteúdo -- note que não é um texto legível como o
JSON era.

## Mini-desafio resolvido

**Desafio:** salvar uma lista de pessoas em JSON e recarregar.

```python
pessoas = [
    {"nome": "Ana", "idade": 28},
    {"nome": "Bruno", "idade": 34},
]

caminho_pessoas = os.path.join(pasta_temporaria, "pessoas.json")

with open(caminho_pessoas, "w", encoding="utf-8") as arquivo:
    json.dump(pessoas, arquivo, indent=2, ensure_ascii=False)

with open(caminho_pessoas, "r", encoding="utf-8") as arquivo:
    pessoas_carregadas = json.load(arquivo)

for pessoa in pessoas_carregadas:
    print(pessoa["nome"], pessoa["idade"])
```

## Limpando o workspace temporário

```python
import shutil

shutil.rmtree(pasta_temporaria)
print("Workspace temporário removido.")
```
